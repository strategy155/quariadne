Let's start with generating a doubly stochatisc matrix!

```python
import numpy as np
import scipy.optimize
rng = np.random.default_rng()
non_square_random = np.pad(rng.random((2,3,)), pad_width=((0,1),(0,0)))

rows, cols = scipy.optimize.linear_sum_assignment(-non_square_random)
rows, cols
```

Now let's doo the S-K algo:

```python
transp = non_square_random.T
sumed = np.sum(transp, axis=0)

a = np.zeros(non_square_random.shape)
np.divide(non_square_random.T, np.sum(non_square_random.T, axis=0),out=a, where= np.sum(non_square_random.T, axis=0)!=0)
a
```
